# **BROWSWER HISTORY -> DLL**

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class BrowserHistory:
    def __init__(self):
        self.current = None

    def visit(self, page):
        new_node = Node(page)
        if self.current:
            self.current.next = None  # Clear forward history
            new_node.prev = self.current
            self.current.next = new_node
        self.current = new_node

    def back(self):
        if self.current and self.current.prev:
            self.current = self.current.prev
        else:
            print("No previous page")

    def forward(self):
        if self.current and self.current.next:
            self.current = self.current.next
        else:
            print("No next page")

    def current_page(self):
        if self.current:
            print("Current page:", self.current.data)
        else:
            print("No Pages Visited yet")

    def show_history(self):
        temp = self.current
        # Move to the start of the history
        while temp and temp.prev:
            temp = temp.prev
        # Print history from the beginning
        while temp:
            if temp == self.current:
                print(f"[{temp.data}] <== current")
            else:
                print(temp.data)
            temp = temp.next

# Test code
bh = BrowserHistory()
bh.visit("google.com")
bh.visit("youtube.com")
bh.visit("amazon.com")
bh.current_page()   # Output: amazon.com

bh.back()
bh.current_page()   # Output: youtube.com

bh.back()
bh.visit("github.com")
bh.current_page()   # Output: github.com

bh.forward()        # Output: No next page
bh.show_history()   # Display all history, marking current page
```

# **STACK USING LIST**

```python
class Stack:
  def __init__(self,max_size):
    self.stack = []
    self.max_size = max_size
  def is_full(self):
    return len(self.stack) >= self.max_size
  def is_empty(self):
    return len(self.stack) == 0

  def push(self,data):
    if self.is_full():
      return "Overflow"
    self.stack.append(data)

  def pop(self):
    if self.is_empty():
      return "Underflow"
    return self.stack.pop()

  def peek(self):
    if self.is_empty():
      return "Stack is empty"
    return self.stack[-1]

  def size(self):
    return len(self.stack)

  def display(self):
    if self.is_empty():
      return "Stack is empty"
    print("Stack elements are")
    for i in reversed(self.stack):
      print(i)
s1 = Stack(5)
s1.push(1)
s1.push(2)
s1.push(3)
s1.push(4)
s1.push(5)
s1.push(6)
print("peeked elements",s1.peek())
print("popped elements",s1.pop())
print("size of stack",s1.size())
print("After popping")
s1.display()
```

# **STACK USING LINKED LIST**

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class Stack:
  def __init__(self, capacity):
    self.top = None
    self.capacity = capacity
    self.size = 0

  def is_empty(self):
    return self.top is None

  def is_full(self):
    return self.size >= self.capacity

  def get_size(self):
    return self.size

  def push(self, dat):
    if self.is_full():
      print("Overflow")
      return
    new_node = Node(dat)
    new_node.next = self.top
    self.top = new_node
    self.size += 1

  def pop(self):
    if self.is_empty():
      return "Underflow"
    popped = self.top.data
    self.top = self.top.next
    self.size -= 1
    return popped

  def peek(self):
    if self.is_empty():
      return "Stack is empty"
    return self.top.data

  def display(self):
    if self.is_empty():
      print("Stack is empty")
      return
    print("Stack elements are:")
    temp = self.top
    while temp:
      print(temp.data)
      temp = temp.next

s = Stack(5)
s.push(10)
s.push(20)
s.push(30)
s.push(40)
s.push(50)
s.display()
print("Top element:", s.peek())
print("Popped:", s.pop())
s.display()
print("Size:", s.get_size())
```

### **INFIX TO POSTFIX**

```python
class Stack:
  def __init__(self, max_size=100):  # default max size to avoid passing explicitly
    self.stack = []
    self.max_size = max_size

  def is_full(self):
    return len(self.stack) >= self.max_size

  def is_empty(self):
    return len(self.stack) == 0

  def push(self, data):
    if self.is_full():
      return "Overflow"
    self.stack.append(data)

  def pop(self):
    if self.is_empty():
      return "Underflow"
    return self.stack.pop()

  def peek(self):
    if self.is_empty():
      return "Stack is empty"
    return self.stack[-1]

  def size(self):
    return len(self.stack)

# Infix to Postfix conversion
def precedence(op):
  precedence_order = {'+':1, '-':1, '*':2, '/':2, '^':3}
  return precedence_order.get(op, 0)

def is_operator(c):
  return c in {'+', '-', '*', '/', '^'}

def infix_to_postfix(expression):
  stack = Stack()  # default size will be 100
  output = []

  for char in expression:
    if char.isalnum():  # operand
      output.append(char)
    elif char == '(':
      stack.push(char)
    elif char == ')':
      while not stack.is_empty() and stack.peek() != '(':
        output.append(stack.pop())
      stack.pop()
    elif is_operator(char):
      while (not stack.is_empty() and is_operator(stack.peek()) and
             precedence(stack.peek()) >= precedence(char)):
        output.append(stack.pop())
      stack.push(char)

  while not stack.is_empty():
    output.append(stack.pop())

  return ''.join(output)

infix = input("Enter infix expression: ")
postfix = infix_to_postfix(infix)
print("Postfix:", postfix)
```

```python
class Queue:
  def __init__(self):
    self.queue = []

  def is_empty(self):
    return len(self.queue) == 0

  def size(self):
    return len(self.queue)

  def enqueue(self, data):
    self.queue.append(data)

  def dequeue(self):
    if self.is_empty():
      return "Queue is empty"
    return self.queue.pop(0)

  def front(self):
    return self.queue[0] if not self.is_empty() else "Queue is empty"

  def rear(self):
    return self.queue[-1] if not self.is_empty() else "Queue is empty"

  def display(self):
    if self.is_empty():
      return "Queue is empty"
    print("Queue elements are:")
    for i in self.queue:
      print(i)


q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
q.display()

print("Size of queue is:",q.size())
print("Front:", q.front())
print("Rear:", q.rear())
print("Dequeue:", q.dequeue())
print("After dequeue")
q.display()
```

# **QUEUE** [Circular Queue]
2. Using Linked list

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class Circular_Q:
  def __init__(self):
    self.front = self.rear = None
    self.size = 0

  def is_empty(self):
    return self.front is None

  def get_front(self):
    return self.front.data if self.front else None

  def get_rear(self):
    return self.rear.data if self.rear else None

  def get_size(self):  # Renamed to avoid conflict
    return self.size

  def enqueue(self, data):
    new_node = Node(data)
    if self.is_empty():
      self.front = self.rear = new_node
      self.rear.next = self.front
    else:
      self.rear.next = new_node
      self.rear = new_node
      self.rear.next = self.front
    self.size += 1

  def dequeue(self):
    if self.is_empty():
      return "Queue is empty"
    data = self.front.data
    if self.front == self.rear:  # Only one node
      self.front = self.rear = None
    else:
      self.front = self.front.next
      self.rear.next = self.front
    self.size -= 1
    return data

  def display(self):
    if self.is_empty():
      print("Queue is empty")
      return
    temp = self.front
    print("Queue elements are:", end=" ")
    while True:
      print(temp.data, end=" -> ")
      temp = temp.next
      if temp == self.front:
        break
    print()


q = Circular_Q()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
q.display()

print("Size of queue is:", q.get_size())
print("Front:", q.get_front())
print("Rear:", q.get_rear())
print("Dequeued:", q.dequeue())
print("After dequeue:")
q.display()
print("Size of queue is:", q.get_size())
```

## **ROUND-ROBIN CPU SCHEDULING**

```python
from collections import deque

class Process:
  def __init__(self, pid, burst_time):
    self.pid = pid
    self.burst_time = burst_time

def round_robin(processes, time_quantum):
  queue = deque(processes)
  time = 0

  while queue:
    process = queue.popleft()

    if process.burst_time > time_quantum:
      print(f"Time {time} : Process {process.pid} runs for {time_quantum} units.")
      time += time_quantum
      process.burst_time -= time_quantum
      queue.append(process)
    else:
      print(f"Time {time} : Process {process.pid} runs for {process.burst_time} units and completes.")
      time += process.burst_time

# Create process list
processes = [Process(1, 10), Process(2, 5), Process(3, 8)]
time_quantum = 2

# Run scheduler
round_robin(processes, time_quantum)
```

# **Traversal Techniques of Graph**

```python
from collections import deque
class GraphMatrix:
  def __init__(self,vertices):
    self.V = vertices
    self.graph = [[0] * vertices for _ in range(vertices)]
  def add_edge(self,u,v):
    self.graph[u][v] = 1
    #undirected graphs
    self.graph[v][u] = 1

  def dfs(self,node,visited = None):
    if visited == None:
      visited = [False]*self.V
      print("DFS: ",end=' ')
    visited[node] = True
    print(node,end=' ')
    for i in range(self.V):
      if self.graph[node][i] == 1 and not visited[i]:
        self.dfs(i,visited)

  def bfs(self,start):
    visited = [False]*self.V
    queue = deque([start])
    visited[start] = True
    print("\nBFS: ",end=' ')
    while queue:
      node = queue.popleft()
      print(node,end=' ')
      for i in range(self.V):
        if self.graph[node][i] == 1 and not visited[i]:
          visited[i] = True
          queue.append(i)


g = GraphMatrix(5)
g.add_edge(0,1)
g.add_edge(0,2)
g.add_edge(1,3)
g.add_edge(1,4)
g.dfs(0)
g.bfs(0)
print("\nAdjacency List:")
for i in range(g.V):
  print(f"{i}:{g.graph[i]}")
```
